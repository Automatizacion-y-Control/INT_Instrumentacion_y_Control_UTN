## 1. Introducción a los sistemas de primer orden

Los sistemas de primer orden son aquellos cuya dinámica puede ser descrita por una ecuación diferencial ordinaria de primer orden. Son los sistemas dinámicos más simples, después de los sistemas estáticos, y son fundamentales para entender conceptos más complejos en el análisis de sistemas.

Se caracterizan por tener un único elemento con capacidad de almacenar energía. Esta característica hace que su respuesta ante un cambio en la entrada no sea instantánea, sino que evolucione gradualmente hacia un nuevo estado de equilibrio.

Ejemplos comunes de sistemas de primer orden en diversas áreas de la ingeniería y la física incluyen:

*   **Sistemas eléctricos:** Un circuito RC (resistencia-condensador) o un circuito RL (resistencia-inductor).
*   **Sistemas térmicos:** La variación de temperatura de un objeto al ser expuesto a un cambio en la temperatura ambiente.
*   **Sistemas mecánicos:** Un sistema de amortiguación viscosa sin masa.
*   **Sistemas de fluidos:** El llenado o vaciado de un tanque con resistencia a la salida.

La comprensión de los sistemas de primer orden es crucial porque muchos sistemas de orden superior pueden aproximarse por sistemas de primer orden bajo ciertas condiciones, o pueden ser analizados como una combinación de sistemas de primer orden.

## 2. Ecuación diferencial

La ecuación diferencial que describe el comportamiento de un sistema lineal invariante en el tiempo de primer orden es:

$$ \tau \frac{dy(t)}{dt} + y(t) = K u(t) $$

Donde:

*   $y(t)$: es la salida del sistema en función del tiempo.
*   $u(t)$: es la entrada del sistema en función del tiempo.
*   $\tau$: es la **constante de tiempo** del sistema, que determina la velocidad de respuesta.
*   $K$: es la **ganancia estática** del sistema, que relaciona el cambio en la salida con un cambio permanente en la entrada una vez alcanzado el estado estacionario.

Esta ecuación nos indica que la suma de la salida y su derivada (ponderada por la constante de tiempo) es proporcional a la entrada. La constante de tiempo ($\tau$) es un parámetro fundamental que caracteriza la rapidez con la que el sistema responde a los cambios en su entrada. Una $\tau$ pequeña indica una respuesta rápida, mientras que una $\tau$ grande indica una respuesta lenta. La ganancia estática ($K$) nos dice cuál será el valor final de la salida ante una entrada constante.

## 3. Función de transferencia

La función de transferencia es una representación en el dominio de Laplace de la relación entre la salida de un sistema y su entrada, asumiendo condiciones iniciales nulas. Es una herramienta muy útil para analizar la dinámica de los sistemas lineales e invariantes en el tiempo.

Para obtener la función de transferencia de un sistema de primer orden, aplicamos la transformada de Laplace a la ecuación diferencial:

$$ \tau \frac{dy(t)}{dt} + y(t) = K u(t) $$

Aplicando la transformada de Laplace (con condiciones iniciales nulas, $y(0)=0$):

$$ \mathcal{L}\left\{ \tau \frac{dy(t)}{dt} + y(t) \right\} = \mathcal{L}\left\{ K u(t) \right\} $$

$$ \tau s Y(s) + Y(s) = K U(s) $$

Factorizando $Y(s)$:

$$ Y(s) (\tau s + 1) = K U(s) $$

La función de transferencia $G(s)$ se define como la relación entre la transformada de Laplace de la salida $Y(s)$ y la transformada de Laplace de la entrada $U(s)$:

$$ G(s) = \frac{Y(s)}{U(s)} = \frac{K}{\tau s + 1} $$

Esta es la forma general de la función de transferencia de un sistema de primer orden. La función de transferencia nos proporciona información sobre los polos y ceros del sistema, que a su vez determinan su estabilidad y respuesta transitoria. Para un sistema de primer orden, hay un único polo en $s = -1/\tau$.

## 4. Respuesta al escalón unitario

La respuesta al escalón unitario es una de las respuestas más importantes para caracterizar un sistema dinámico. Consiste en aplicar una entrada que cambia instantáneamente de 0 a 1 en el tiempo $t=0$ (la función escalón unitario, denotada como $u(t) = 1$ para $t \ge 0$ y $u(t) = 0$ para $t < 0$) y observar cómo evoluciona la salida del sistema.

Para un sistema de primer orden con función de transferencia $G(s) = \frac{K}{\tau s + 1}$, la transformada de Laplace de la respuesta al escalón unitario es:

$$ Y(s) = G(s) U(s) $$

Donde $U(s) = \mathcal{L}\{u(t)\} = \frac{1}{s}$ para una entrada de escalón unitario.

$$ Y(s) = \frac{K}{\tau s + 1} \cdot \frac{1}{s} $$

Para encontrar la respuesta en el dominio del tiempo $y(t)$, aplicamos la transformada inversa de Laplace. Podemos usar fracciones parciales para simplificar la expresión:

$$ Y(s) = \frac{A}{s} + \frac{B}{\tau s + 1} = \frac{A(\tau s + 1) + Bs}{s(\tau s + 1)} = \frac{(A\tau + B)s + A}{s(\tau s + 1)} $$

Comparando los numeradores:
$K = (A\tau + B)s + A$

Igualando los coeficientes de $s$: $A\tau + B = 0$
Igualando los términos constantes: $A = K$

Sustituyendo $A=K$ en la primera ecuación: $K\tau + B = 0 \implies B = -K\tau$

Por lo tanto,
$$ Y(s) = \frac{K}{s} - \frac{K\tau}{\tau s + 1} = \frac{K}{s} - \frac{K}{s + 1/\tau} $$

Aplicando la transformada inversa de Laplace:
$$ y(t) = \mathcal{L}^{-1}\left\{ \frac{K}{s} \right\} - \mathcal{L}^{-1}\left\{ \frac{K}{s + 1/\tau} \right\} $$

$$ y(t) = K u(t) - K e^{-t/\tau} u(t) $$

Para $t \ge 0$, donde $u(t) = 1$:

$$ y(t) = K (1 - e^{-t/\tau}) $$

Esta ecuación describe la respuesta de un sistema de primer orden a un escalón unitario. La salida $y(t)$ comienza en 0 y se aproxima asintóticamente al valor de estado estacionario $K$ a medida que el tiempo $t$ aumenta. La velocidad a la que la salida se acerca a $K$ está determinada por la constante de tiempo $\tau$.

Ahora, simulemos esta respuesta en Python.

```python
# Para este código, necesitaremos la librería 'control'.
!pip install control numpy matplotlib
```

```python
# Para este código, necesitaremos la librería 'control'. Si no la tienes instalada, descomenta la siguiente línea y ejecútala:
# !pip install control numpy matplotlib

import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Definir los parámetros del sistema
K = 1.0  # Ganancia estática
tau = 1.0 # Constante de tiempo

# Crear la función de transferencia del sistema de primer orden
# G(s) = K / (tau*s + 1)
num = [K]
den = [tau, 1]
sistema_primer_orden = ct.tf(num, den)

print("Función de transferencia del sistema de primer orden:")
print(sistema_primer_orden)

# Definir el vector de tiempo para la simulación
# Simular durante aproximadamente 5 * tau para ver la respuesta completa
tiempo = np.linspace(0, 5 * tau, 500)

# Obtener la respuesta al escalón unitario
# step_response devuelve el tiempo, la respuesta
tiempo_respuesta, respuesta_escalon = ct.step_response(sistema_primer_orden, tiempo)

# Graficar la respuesta al escalón
plt.figure()
plt.plot(tiempo_respuesta, respuesta_escalon)
plt.title('Respuesta al Escalón Unitario de un Sistema de Primer Orden')
plt.xlabel('Tiempo [unidades]')
plt.ylabel('Salida y(t)')
plt.grid(True)
plt.axhline(y=K, color='r', linestyle='--', label=f'Valor de estado estacionario (K={K})')
plt.axvline(x=tau, color='g', linestyle='--', label=f'Tiempo = tau ({tau})')
# Agregar una línea para el 63.2% del valor final en t = tau
plt.axhline(y=K * (1 - np.exp(-1)), color='purple', linestyle=':', label=f'63.2% de K en t=tau ({K * (1 - np.exp(-1)):.2f})')

plt.legend()
plt.show()

# Mostrar algunos valores clave de la respuesta
print("\nValores clave de la respuesta:")
print(f"En t=0: y(0) = {respuesta_escalon[0]:.4f}")
print(f"En t=tau ({tau}): y({tau}) = {respuesta_escalon[np.argmin(np.abs(tiempo_respuesta - tau))]:.4f}")
print(f"Valor de estado estacionario (aproximado al final de la simulación): y({tiempo_respuesta[-1]:.2f}) = {respuesta_escalon[-1]:.4f}")
```

## 5. Constante de tiempo

La **constante de tiempo** ($\tau$) es un parámetro fundamental en los sistemas de primer orden que determina la velocidad de su respuesta transitoria. Se define como el tiempo que tarda la salida del sistema en alcanzar aproximadamente el 63.2% de su valor final (estado estacionario) ante una entrada de escalón unitario, asumiendo condiciones iniciales nulas.

En la ecuación de respuesta al escalón unitario $y(t) = K (1 - e^{-t/\tau})$, cuando $t = \tau$:

$$ y(\tau) = K (1 - e^{-\tau/\tau}) = K (1 - e^{-1}) \approx K (1 - 0.3678) \approx 0.632 K $$

Esto significa que cuanto menor sea el valor de $\tau$, más rápido responderá el sistema a los cambios en la entrada, ya que alcanzará el 63.2% de su valor final en menos tiempo. Por el contrario, un valor grande de $\tau$ indica un sistema más lento.

Otros puntos de referencia importantes relacionados con la constante de tiempo son:

*   En $t = 2\tau$, la salida ha alcanzado aproximadamente el 86.5% de su valor final.
*   En $t = 3\tau$, la salida ha alcanzado aproximadamente el 95% de su valor final.
*   En $t = 4\tau$, la salida ha alcanzado aproximadamente el 98.2% de su valor final.
*   En $t = 5\tau$, la salida ha alcanzado aproximadamente el 99.3% de su valor final. A menudo, se considera que después de $5\tau$ el sistema ha alcanzado prácticamente el estado estacionario.

La constante de tiempo es una medida directa de la inercia del sistema. En un circuito RC, $\tau = RC$. En un circuito RL, $\tau = L/R$. Comprender la constante de tiempo es crucial para predecir y diseñar el comportamiento dinámico de los sistemas de primer orden.

## 6. Ejemplos prácticos

Los sistemas de primer orden se manifiestan en diversas disciplinas de la ingeniería y la física. Su estudio nos permite comprender el comportamiento dinámico de sistemas aparentemente diferentes bajo un marco matemático común. A continuación, exploraremos cómo los principios de los sistemas de primer orden se aplican en algunas áreas clave:

*   **Sistemas Eléctricos:** Circuitos simples que involucran resistencias y capacitores (circuitos RC) o resistencias e inductores (circuitos RL) exhiben un comportamiento de primer orden. La carga o descarga de un capacitor, o el establecimiento de corriente en un inductor, son procesos gobernados por ecuaciones diferenciales de primer orden.

*   **Sistemas Térmicos:** La transferencia de calor entre objetos o entre un objeto y su entorno, bajo ciertas condiciones, puede modelarse como un sistema de primer orden. Un ejemplo clásico es cómo la temperatura de un objeto se aproxima a la temperatura ambiente con el tiempo.

*   **Sistemas Mecánicos:** Aunque los sistemas mecánicos a menudo son de segundo orden (debido a la inercia y el almacenamiento de energía potencial), algunos casos simplificados, como un sistema con amortiguación viscosa y sin masa significativa, pueden aproximarse como de primer orden.

*   **Sistemas de Fluidos:** El llenado o vaciado de tanques donde la resistencia al flujo es proporcional a la altura del líquido es otro ejemplo. La tasa de cambio del nivel del líquido depende linealmente del nivel actual.

Comprender cómo identificar y modelar estos sistemas como de primer orden nos permite predecir su respuesta temporal y diseñar sistemas de control adecuados. A continuación, veremos algunos ejemplos específicos con su modelado y simulación.

### Ejemplo Eléctrico: Circuito RC

Consideremos un circuito RC simple en serie, con una resistencia $R$ y un condensador $C$, conectado a una fuente de voltaje de entrada $V_{in}(t)$. La salida que nos interesa es el voltaje a través del condensador, $V_{out}(t)$.

Aplicando la Ley de Voltaje de Kirchhoff, la suma de las caídas de voltaje a través de la resistencia y el condensador debe ser igual al voltaje de entrada:

$$ V_R(t) + V_C(t) = V_{in}(t) $$

Sabemos que el voltaje a través de la resistencia es $V_R(t) = I(t) R$, donde $I(t)$ es la corriente que fluye por el circuito. La corriente a través del condensador es $I(t) = C \frac{dV_{out}(t)}{dt}$. Como el condensador y la resistencia están en serie, la misma corriente fluye a través de ambos, por lo tanto, $I(t)$ es la corriente total del circuito.

Sustituyendo la expresión de la corriente en la ecuación de voltaje de la resistencia:

$$ V_R(t) = \left(C \frac{dV_{out}(t)}{dt}\right) R = RC \frac{dV_{out}(t)}{dt} $$

Ahora, sustituimos $V_R(t)$ y $V_C(t) = V_{out}(t)$ en la ecuación de Kirchhoff:

$$ RC \frac{dV_{out}(t)}{dt} + V_{out}(t) = V_{in}(t) $$

Esta ecuación es exactamente la forma de la ecuación diferencial de un sistema de primer orden:

$$ \tau \frac{dy(t)}{dt} + y(t) = K u(t) $$

Comparando las dos ecuaciones, podemos identificar:

*   La salida $y(t)$ es $V_{out}(t)$.
*   La entrada $u(t)$ es $V_{in}(t)$.
*   La constante de tiempo $\tau$ es $RC$.
*   La ganancia estática $K$ es 1 (asumiendo que la entrada es un voltaje y la salida es un voltaje, la ganancia es unitaria en estado estacionario).

La función de transferencia para este circuito RC es:

$$ G(s) = \frac{V_{out}(s)}{V_{in}(s)} = \frac{1}{RCs + 1} $$

Donde $\tau = RC$.

Ahora, vamos a simular la respuesta de este circuito a una entrada de escalón de voltaje (por ejemplo, conectar el circuito a una batería de voltaje constante en $t=0$).

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Parámetros del circuito RC
R = 1000  # Resistencia en Ohms
C = 100e-6 # Capacitancia en Faradios (100 microFaradios)

# Calcular la constante de tiempo
tau_rc = R * C
print(f"La constante de tiempo (tau) del circuito RC es: {tau_rc:.4f} segundos")

# Ganancia estática (para voltaje de salida con entrada de voltaje)
K_rc = 1.0

# Crear la función de transferencia del circuito RC
# G(s) = 1 / (tau*s + 1)
num_rc = [K_rc]
den_rc = [tau_rc, 1]
sistema_rc = ct.tf(num_rc, den_rc)

print("\nFunción de transferencia del circuito RC:")
print(sistema_rc)

# Definir el vector de tiempo para la simulación
# Simular durante aproximadamente 5 * tau_rc
tiempo_rc = np.linspace(0, 5 * tau_rc, 500)

# Simular la respuesta a un escalón unitario de voltaje (entrada de 1V)
tiempo_respuesta_rc, respuesta_escalon_rc = ct.step_response(sistema_rc, tiempo_rc)

# Graficar la respuesta al escalón
plt.figure()
plt.plot(tiempo_respuesta_rc, respuesta_escalon_rc)
plt.title('Respuesta al Escalón de Voltaje en un Circuito RC')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Voltaje en el Condensador [V]')
plt.grid(True)
plt.axhline(y=K_rc, color='r', linestyle='--', label=f'Valor final (K={K_rc}V)')
plt.axvline(x=tau_rc, color='g', linestyle='--', label=f'Tiempo = tau ({tau_rc:.4f}s)')
# Agregar una línea para el 63.2% del valor final en t = tau
plt.axhline(y=K_rc * (1 - np.exp(-1)), color='purple', linestyle=':', label=f'63.2% de K en t=tau ({K_rc * (1 - np.exp(-1)):.2f}V)')

plt.legend()
plt.show()

# Mostrar algunos valores clave de la respuesta
print("\nValores clave de la respuesta del circuito RC:")
print(f"En t=0: V_out(0) = {respuesta_escalon_rc[0]:.4f} V")
print(f"En t=tau ({tau_rc:.4f}): V_out({tau_rc:.4f}) = {respuesta_escalon_rc[np.argmin(np.abs(tiempo_respuesta_rc - tau_rc))]:.4f} V")
print(f"Valor de estado estacionario (aproximado al final de la simulación): V_out({tiempo_respuesta_rc[-1]:.4f}) = {respuesta_escalon_rc[-1]:.4f} V")
```

### Ejemplo Térmico: Calentamiento/Enfriamiento de un Objeto

Consideremos un objeto con masa $m$ y calor específico $c$, inmerso en un ambiente a temperatura $T_a$. Asumimos que la transferencia de calor entre el objeto y el ambiente es convectiva, gobernada por un coeficiente de transferencia de calor $h$ y el área de superficie $A$ del objeto. La temperatura del objeto en el tiempo $t$ es $T(t)$.

Según la Ley de Enfriamiento de Newton, la tasa de cambio de la temperatura del objeto es proporcional a la diferencia de temperatura entre el objeto y el ambiente:

$$ \frac{dQ}{dt} = -hA (T(t) - T_a) $$

Donde $\frac{dQ}{dt}$ es la tasa de transferencia de calor. También sabemos que la cantidad de calor en el objeto es $Q = mc (T(t) - T_{ref})$, donde $T_{ref}$ es una temperatura de referencia constante. Por lo tanto, la tasa de cambio de calor también se relaciona con la tasa de cambio de temperatura:

$$ \frac{dQ}{dt} = mc \frac{dT(t)}{dt} $$

Igualando las dos expresiones para $\frac{dQ}{dt}$:

$$ mc \frac{dT(t)}{dt} = -hA (T(t) - T_a) $$

Reorganizando la ecuación para que se parezca a la forma estándar de primer orden:

$$ mc \frac{dT(t)}{dt} + hA T(t) = hA T_a $$

Dividiendo por $hA$:

$$ \frac{mc}{hA} \frac{dT(t)}{dt} + T(t) = T_a $$

Comparando con la ecuación general de primer orden $\tau \frac{dy(t)}{dt} + y(t) = K u(t)$:

*   La salida $y(t)$ es la temperatura del objeto $T(t)$.
*   La entrada $u(t)$ es la temperatura ambiente $T_a$.
*   La constante de tiempo $\tau$ es $\frac{mc}{hA}$.
*   La ganancia estática $K$ es 1 (ya que en estado estacionario, $T(t) = T_a$).

Para simular la respuesta a un cambio escalón en la temperatura ambiente (por ejemplo, mover el objeto de un ambiente a otro con diferente temperatura), podemos considerar que la entrada $T_a$ cambia instantáneamente.

La función de transferencia, si consideramos la desviación de la temperatura ambiente $\Delta T(t) = T(t) - T_a$ como salida y la desviación de la temperatura ambiente $\Delta T_a(t)$ como entrada, es:

$$ G(s) = \frac{\Delta T(s)}{\Delta T_a(s)} = \frac{1}{\tau s + 1} $$

Donde $\tau = \frac{mc}{hA}$. Sin embargo, a menudo es más directo simular la ecuación diferencial original.

Vamos a simular la respuesta de la temperatura del objeto cuando la temperatura ambiente cambia abruptamente (un escalón).

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Parámetros del sistema térmico
m = 1.0   # Masa del objeto [kg]
c = 4186  # Calor específico del agua [J/kg°C]
h = 10    # Coeficiente de transferencia de calor [W/m²°C]
A = 0.1   # Área de superficie del objeto [m²]

# Temperatura ambiente inicial y final (simulando un escalón)
Ta_inicial = 20.0 # °C
Ta_final = 80.0   # °C
delta_Ta = Ta_final - Ta_inicial # Cambio escalón en la temperatura ambiente

# Calcular la constante de tiempo
tau_termico = (m * c) / (h * A)
print(f"La constante de tiempo (tau) del sistema térmico es: {tau_termico:.4f} segundos")

# Definir la función de transferencia considerando la desviación de la temperatura ambiente como entrada
# y la desviación de la temperatura del objeto como salida.
# G(s) = 1 / (tau*s + 1)
num_termico = [1.0] # Ganancia estática es 1 para la desviación
den_termico = [tau_termico, 1]
sistema_termico_desviacion = ct.tf(num_termico, den_termico)

print("\nFunción de transferencia (en términos de desviación) del sistema térmico:")
print(sistema_termico_desviacion)

# Definir el vector de tiempo para la simulación
# Simular durante aproximadamente 5 * tau_termico
tiempo_termico = np.linspace(0, 5 * tau_termico, 500)

# Simular la respuesta a un escalón en la desviación de la temperatura ambiente
# La entrada escalón es de magnitud delta_Ta
tiempo_respuesta_termico_desviacion, respuesta_escalon_termico_desviacion = ct.step_response(delta_Ta * sistema_termico_desviacion, tiempo_termico)

# La temperatura real del objeto es la desviación + la temperatura ambiente inicial
respuesta_escalon_termico_real = respuesta_escalon_termico_desviacion + Ta_inicial

# Graficar la respuesta al escalón
plt.figure()
plt.plot(tiempo_respuesta_termico_desviacion, respuesta_escalon_termico_real)
plt.title('Respuesta de Temperatura de Objeto a Cambio Escalón en Temperatura Ambiente')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Temperatura del Objeto [°C]')
plt.grid(True)
plt.axhline(y=Ta_final, color='r', linestyle='--', label=f'Temperatura final ({Ta_final}°C)')
plt.axvline(x=tau_termico, color='g', linestyle='--', label=f'Tiempo = tau ({tau_termico:.4f}s)')
# Agregar una línea para el 63.2% del cambio total desde la temperatura inicial
temp_en_tau = Ta_inicial + (Ta_final - Ta_inicial) * (1 - np.exp(-1))
plt.axhline(y=temp_en_tau, color='purple', linestyle=':', label=f'Temperatura en t=tau ({temp_en_tau:.2f}°C)')


plt.legend()
plt.show()

# Mostrar algunos valores clave de la respuesta
print("\nValores clave de la respuesta de temperatura:")
print(f"En t=0: T(0) = {respuesta_escalon_termico_real[0]:.4f} °C")
print(f"En t=tau ({tau_termico:.4f}): T({tau_termico:.4f}) = {respuesta_escalon_termico_real[np.argmin(np.abs(tiempo_respuesta_termico_desviacion - tau_termico))]:.4f} °C")
print(f"Valor de estado estacionario (aproximado al final de la simulación): T({tiempo_respuesta_termico_desviacion[-1]:.4f}) = {respuesta_escalon_termico_real[-1]:.4f} °C")
```

### Ejemplo Mecánico: Sistema de Amortiguación Viscosa

Consideremos un sistema mecánico simple que consiste en un objeto moviéndose en un medio viscoso (como un líquido espeso o aire denso) sin inercia significativa (masa despreciable) y sin elementos elásticos (resortes). La única fuerza que se opone al movimiento es la fuerza de amortiguación viscosa, que es proporcional a la velocidad del objeto.

Según la ley de amortiguación viscosa, la fuerza de amortiguación ($F_v$) está dada por:

$$ F_v(t) = b v(t) $$

Donde:

*   $b$: es el coeficiente de amortiguación viscosa.
*   $v(t)$: es la velocidad del objeto en el tiempo $t$.

Supongamos que aplicamos una fuerza externa $F_{in}(t)$ al objeto. Como la masa es despreciable, la suma de las fuerzas debe ser cero (aproximadamente, según la segunda ley de Newton, $F = ma$, si $m \approx 0$, entonces $F \approx 0$). Por lo tanto, la fuerza aplicada debe ser igual a la fuerza de amortiguación:

$$ F_{in}(t) = F_v(t) $$

$$ F_{in}(t) = b v(t) $$

Si la salida que nos interesa es la velocidad $v(t)$, podemos reorganizar la ecuación:

$$ b v(t) = F_{in}(t) $$

Para que se parezca a la forma estándar de primer orden $\tau \frac{dy(t)}{dt} + y(t) = K u(t)$, necesitamos una derivada de la salida. Si consideramos la posición $x(t)$ como salida, donde $v(t) = \frac{dx(t)}{dt}$, la ecuación se convierte en:

$$ b \frac{dx(t)}{dt} = F_{in}(t) $$

Dividiendo por $b$:

$$ \frac{dx(t)}{dt} = \frac{1}{b} F_{in}(t) $$

Esta es una ecuación diferencial de primer orden en la posición $x(t)$. Comparando con la forma general $\tau \frac{dy(t)}{dt} + y(t) = K u(t)$:

*   La salida $y(t)$ es la posición $x(t)$.
*   La entrada $u(t)$ es la fuerza aplicada $F_{in}(t)$.
*   El término $y(t)$ en la ecuación general está ausente aquí. Esto significa que la constante de tiempo es infinita en esta forma simple, lo que indica que el sistema no alcanza un estado estacionario en posición ante una entrada constante de fuerza (la velocidad sería constante, y la posición seguiría aumentando).

Sin embargo, es más común considerar la **velocidad** $v(t)$ como la salida para este tipo de sistema de primer orden. Revisitando la ecuación:

$$ b v(t) = F_{in}(t) $$

Esta ecuación puede verse como un caso límite de la forma general si consideramos una "constante de tiempo inversa" o si la reescribimos ligeramente. Una mejor manera de ver esto como primer orden es considerar la dinámica de la velocidad. Aunque la ecuación no tiene el término $y(t)$, podemos pensar en ella en el dominio de Laplace:

$$ b V(s) = F_{in}(s) $$

La función de transferencia, si la salida es la velocidad $v(t)$ y la entrada es la fuerza $F_{in}(t)$, es:

$$ G(s) = \frac{V(s)}{F_{in}(s)} = \frac{1}{b} $$

Esta es una función de transferencia de orden cero (un sistema estático) si la consideramos estrictamente en el dominio de Laplace con esa definición de entrada/salida.

**Sin embargo**, muchos textos y contextos consideran sistemas con amortiguación viscosa y masa *no despreciable* como sistemas de SEGUNDO orden. Para obtener un sistema de PRIMER orden con amortiguación, a menudo se considera la relación entre la fuerza aplicada y la VELOCIDAD, o la relación entre el desplazamiento y la fuerza en un sistema con amortiguación y un resorte (ignorando la masa, lo que nos devuelve a la primera ecuación).

**Una mejor analogía de primer orden mecánico** es un sistema con un resorte y un amortiguador en paralelo, con una fuerza aplicada, e interesándonos en la fuerza en el amortiguador o resorte. O, más comúnmente, sistemas donde la fuerza de entrada es proporcional a la VELOCIDAD de algún componente, resultando en una ecuación de primer orden en el desplazamiento.

**Volviendo al sistema de amortiguación viscosa simple (sin masa ni resorte):** Si consideramos la fuerza aplicada $F_{in}(t)$ como entrada y la **velocidad** $v(t)$ como salida, la relación es instantánea ($v(t) = \frac{1}{b} F_{in}(t)$), que es un comportamiento de orden cero.

**Para tener un verdadero sistema de primer orden mecánico**, necesitamos almacenamiento de energía en UN solo elemento. Esto ocurre con un resorte (almacena energía potencial) o un amortiguador viscoso si consideramos la fuerza como entrada y la posición como salida (la energía se disipa, pero la ecuación es de primer orden en posición).

Consideremos nuevamente el sistema con amortiguación viscosa y un resorte en paralelo, pero donde la masa es despreciable. La fuerza aplicada $F_{in}(t)$ se opone a la fuerza del resorte ($K_s x(t)$, donde $K_s$ es la constante del resorte y $x(t)$ es el desplazamiento) y la fuerza del amortiguador ($b v(t) = b \frac{dx(t)}{dt}$).

$$ F_{in}(t) = K_s x(t) + b \frac{dx(t)}{dt} $$

Reorganizando:

$$ b \frac{dx(t)}{dt} + K_s x(t) = F_{in}(t) $$

Dividiendo por $K_s$:

$$ \frac{b}{K_s} \frac{dx(t)}{dt} + x(t) = \frac{1}{K_s} F_{in}(t) $$

¡Ahora sí tenemos la forma de primer orden!

$$ \tau \frac{dy(t)}{dt} + y(t) = K u(t) $$

Comparando:

*   La salida $y(t)$ es el desplazamiento $x(t)$.
*   La entrada $u(t)$ es la fuerza aplicada $F_{in}(t)$.
*   La constante de tiempo $\tau$ es $\frac{b}{K_s}$.
*   La ganancia estática $K$ es $\frac{1}{K_s}$.

Este sistema representa un modelo simplificado, por ejemplo, de ciertos tipos de montajes antivibración o la respuesta de materiales viscoelásticos a cargas.

Vamos a simular la respuesta de este sistema mecánico simplificado (amortiguador y resorte sin masa) a una fuerza de escalón unitario (aplicar una fuerza constante repentinamente).

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Parámetros del sistema mecánico (amortiguador y resorte sin masa)
b = 10.0 # Coeficiente de amortiguación viscosa [Ns/m]
Ks = 5.0 # Constante del resorte [N/m]

# Calcular la constante de tiempo
tau_mecanico = b / Ks
print(f"La constante de tiempo (tau) del sistema mecánico es: {tau_mecanico:.4f} segundos")

# Calcular la ganancia estática
K_mecanico = 1 / Ks
print(f"La ganancia estática (K) del sistema mecánico es: {K_mecanico:.4f} m/N")


# Crear la función de transferencia del sistema mecánico
# G(s) = K / (tau*s + 1)
num_mecanico = [K_mecanico]
den_mecanico = [tau_mecanico, 1]
sistema_mecanico = ct.tf(num_mecanico, den_mecanico)

print("\nFunción de transferencia del sistema mecánico (desplazamiento vs. fuerza):")
print(sistema_mecanico)

# Definir el vector de tiempo para la simulación
# Simular durante aproximadamente 5 * tau_mecanico
tiempo_mecanico = np.linspace(0, 5 * tau_mecanico, 500)

# Simular la respuesta a un escalón unitario de fuerza (entrada de 1N)
tiempo_respuesta_mecanico, respuesta_escalon_mecanico = ct.step_response(sistema_mecanico, tiempo_mecanico)

# Graficar la respuesta al escalón
plt.figure()
plt.plot(tiempo_respuesta_mecanico, respuesta_escalon_mecanico)
plt.title('Respuesta al Escalón de Fuerza en Sistema Mecánico (Amortiguador-Resorte sin Masa)')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Desplazamiento [m]')
plt.grid(True)
plt.axhline(y=K_mecanico, color='r', linestyle='--', label=f'Valor de estado estacionario (K={K_mecanico:.4f}m)')
plt.axvline(x=tau_mecanico, color='g', linestyle='--', label=f'Tiempo = tau ({tau_mecanico:.4f}s)')
# Agregar una línea para el 63.2% del valor final en t = tau
plt.axhline(y=K_mecanico * (1 - np.exp(-1)), color='purple', linestyle=':', label=f'63.2% de K en t=tau ({K_mecanico * (1 - np.exp(-1)):.4f}m)')

plt.legend()
plt.show()

# Mostrar algunos valores clave de la respuesta
print("\nValores clave de la respuesta de desplazamiento:")
print(f"En t=0: x(0) = {respuesta_escalon_mecanico[0]:.4f} m")
print(f"En t=tau ({tau_mecanico:.4f}): x({tau_mecanico:.4f}) = {respuesta_escalon_mecanico[np.argmin(np.abs(tiempo_respuesta_mecanico - tau_mecanico))]:.4f} m")
print(f"Valor de estado estacionario (aproximado al final de la simulación): x({tiempo_respuesta_mecanico[-1]:.4f}) = {respuesta_escalon_mecanico[-1]:.4f} m")
```

### Ejemplo de Fluidos: Llenado de un Tanque

Consideremos un tanque cilíndrico con área de sección transversal $A_{tanque}$, con un orificio de salida en el fondo con resistencia al flujo. Asumimos que el flujo a través del orificio es linealmente proporcional a la altura del líquido en el tanque (válido para flujos laminares o aproximado para otros regímenes si la resistencia es significativa). La entrada al sistema es el flujo volumétrico de entrada $Q_{in}(t)$, y la salida es la altura del líquido en el tanque $h(t)$.

La ley de conservación de la masa para el líquido en el tanque establece que la tasa de cambio del volumen de líquido es igual al flujo de entrada menos el flujo de salida. El volumen de líquido en el tanque es $V(t) = A_{tanque} h(t)$.

$$ \frac{dV(t)}{dt} = Q_{in}(t) - Q_{out}(t) $$

La tasa de cambio del volumen también es:

$$ \frac{dV(t)}{dt} = A_{tanque} \frac{dh(t)}{dt} $$

Asumiendo que el flujo de salida $Q_{out}(t)$ es linealmente proporcional a la altura $h(t)$:

$$ Q_{out}(t) = \frac{1}{R_f} h(t) $$

Donde $R_f$ es la resistencia al flujo del orificio de salida.

Sustituyendo estas expresiones en la ecuación de conservación de masa:

$$ A_{tanque} \frac{dh(t)}{dt} = Q_{in}(t) - \frac{1}{R_f} h(t) $$

Reorganizando para obtener la forma estándar de primer orden:

$$ A_{tanque} \frac{dh(t)}{dt} + \frac{1}{R_f} h(t) = Q_{in}(t) $$

Dividiendo por $\frac{1}{R_f}$:

$$ A_{tanque} R_f \frac{dh(t)}{dt} + h(t) = R_f Q_{in}(t) $$

Comparando con la ecuación general de primer orden $\tau \frac{dy(t)}{dt} + y(t) = K u(t)$:

*   La salida $y(t)$ es la altura del líquido $h(t)$.
*   La entrada $u(t)$ es el flujo de entrada $Q_{in}(t)$.
*   La constante de tiempo $\tau$ es $A_{tanque} R_f$.
*   La ganancia estática $K$ es $R_f$.

Este modelo es una simplificación (ignoramos efectos inerciales del fluido, asumimos flujo lineal, etc.), pero es útil para entender el comportamiento básico de llenado o vaciado de tanques.

La función de transferencia para este sistema es:

$$ G(s) = \frac{H(s)}{Q_{in}(s)} = \frac{R_f}{A_{tanque} R_f s + 1} $$

Donde $\tau = A_{tanque} R_f$ y $K = R_f$.

Vamos a simular la respuesta de la altura del líquido en el tanque a una entrada de flujo de escalón unitario (por ejemplo, abrir una válvula para tener un flujo de entrada constante a partir de $t=0$).

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Parámetros del sistema de fluidos (Tanque)
A_tanque = 2.0  # Área de sección transversal del tanque [m²]
Rf = 5.0      # Resistencia al flujo de salida [s/m²] (ej: (altura / flujo))

# Calcular la constante de tiempo
tau_fluido = A_tanque * Rf
print(f"La constante de tiempo (tau) del sistema de fluidos es: {tau_fluido:.4f} segundos")

# Calcular la ganancia estática
K_fluido = Rf
print(f"La ganancia estática (K) del sistema de fluidos es: {K_fluido:.4f} m/(m³/s)") # Unidades de altura por unidad de flujo

# Crear la función de transferencia del sistema de fluidos
# G(s) = K / (tau*s + 1)
num_fluido = [K_fluido]
den_fluido = [tau_fluido, 1]
sistema_fluido = ct.tf(num_fluido, den_fluido)

print("\nFunción de transferencia del sistema de fluidos (altura vs. flujo de entrada):")
print(sistema_fluido)

# Definir el vector de tiempo para la simulación
# Simular durante aproximadamente 5 * tau_fluido
tiempo_fluido = np.linspace(0, 5 * tau_fluido, 500)

# Simular la respuesta a un escalón unitario de flujo de entrada (entrada de 1 m³/s)
tiempo_respuesta_fluido, respuesta_escalon_fluido = ct.step_response(sistema_fluido, tiempo_fluido)

# Graficar la respuesta al escalón
plt.figure()
plt.plot(tiempo_respuesta_fluido, respuesta_escalon_fluido)
plt.title('Respuesta al Escalón de Flujo en Sistema de Tanque')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Altura del Líquido [m]')
plt.grid(True)
plt.axhline(y=K_fluido, color='r', linestyle='--', label=f'Valor de estado estacionario (K={K_fluido:.4f}m)')
plt.axvline(x=tau_fluido, color='g', linestyle='--', label=f'Tiempo = tau ({tau_fluido:.4f}s)')
# Agregar una línea para el 63.2% del valor final en t = tau
plt.axhline(y=K_fluido * (1 - np.exp(-1)), color='purple', linestyle=':', label=f'63.2% de K en t=tau ({K_fluido * (1 - np.exp(-1)):.4f}m)')

plt.legend()
plt.show()

# Mostrar algunos valores clave de la respuesta
print("\nValores clave de la respuesta de altura:")
print(f"En t=0: h(0) = {respuesta_escalon_fluido[0]:.4f} m")
print(f"En t=tau ({tau_fluido:.4f}): h({tau_fluido:.4f}) = {respuesta_escalon_fluido[np.argmin(np.abs(tiempo_respuesta_fluido - tau_fluido))]:.4f} m")
print(f"Valor de estado estacionario (aproximado al final de la simulación): h({tiempo_respuesta_fluido[-1]:.4f}) = {respuesta_escalon_fluido[-1]:.4f} m")
```
